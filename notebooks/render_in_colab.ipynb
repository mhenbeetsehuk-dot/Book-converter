# Book-converter: free Colab renderer
Run each cell in order. This uses Colab's CPU unless you select an available GPU. Free sessions can expire. Upload a Blender Python script produced by the Book-converter website.


```
!sudo apt-get update -qq && sudo apt-get install -y -qq blender
```

```
from google.colab import files
uploaded = files.upload()
assert len(uploaded) == 1, 'Upload exactly one Blender .py script'
name = next(iter(uploaded))
assert name.endswith('.py'), 'Only .py scripts are supported'
print('Uploaded:', name)
```

**Security:** Blender Python scripts execute arbitrary code. Only upload scripts you trust and have reviewed. The generated script saves its output to /content/book_converter_scene.mp4.


```
import subprocess
subprocess.run(['blender', '-b', '--python', name], check=True, timeout=3600)
files.download('/content/book_converter_scene.mp4')
```
